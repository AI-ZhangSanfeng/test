# 顺序运行 10 个样本（每个 notebook 独立 kernel）

请下载并解压完整 ZIP，在原 NicheCompass 环境运行。每个样本继承自己原有 GPU 设置，本控制文件使用 nbclient，缺失时可在当前环境安装 `pip install nbclient nbformat`，或直接逐份运行样本 notebook。一个样本失败不会阻止其他样本；执行后的 notebook 与错误日志保存在新输出目录。


```python
from pathlib import Path
import json
import traceback
import nbformat
from nbclient import NotebookClient
NOTEBOOK_NAMES = ['01_Human_breast_cancer_ViHBC_Component_Experiment.ipynb', '02_Human_Prostate_Cancer_Component_Experiment.ipynb', '03_B1_Component_Experiment.ipynb', '04_LC1_Component_Experiment.ipynb', '05_LC5_Component_Experiment.ipynb', '06_SN048_A121573_Rep1_Component_Experiment.ipynb', '07_SN048_A121573_Rep2_Component_Experiment.ipynb', '08_SN84_A120838_Rep1_Component_Experiment.ipynb', '09_SN123_A595688_Rep1_Component_Experiment.ipynb', '10_SN124_A595688_Rep2_Component_Experiment.ipynb']
OUTPUT_ROOT = Path('/home/zhangjunyi/xiangmu/nichecompass-main/outputs/case/TuMeNiche_Component_Contribution')

KERNEL_NAME = "nichecompass"
NOTEBOOK_DIR = Path(
    "/home/zhangjunyi/xiangmu/nichecompass-main/TuMeNiche/src/my/case/"
    "TuMeNiche_Component_Contribution/notebooks"
)

import sys
from jupyter_client.kernelspec import KernelSpecManager

print("当前内核", sys.executable)
print("批量子内核", KernelSpecManager().get_kernel_spec(KERNEL_NAME).argv[0])

EXECUTED_DIR = OUTPUT_ROOT / "_executed_notebooks"
EXECUTED_DIR.mkdir(parents=True, exist_ok=True)
```

```python
batch_status = []
for name in NOTEBOOK_NAMES:
    print("Running", name, flush=True)
    nb = nbformat.read(NOTEBOOK_DIR / name, as_version=4)
    try:
        NotebookClient(nb, timeout=7200, kernel_name=KERNEL_NAME, resources={"metadata": {"path": str(NOTEBOOK_DIR)}}).execute()
        batch_status.append({"notebook": name, "status": "completed"})
    except Exception as error:
        batch_status.append({"notebook": name, "status": "failed", "error": str(error)})
        (EXECUTED_DIR / (name + ".error.txt")).write_text(traceback.format_exc())
    finally:
        nbformat.write(nb, EXECUTED_DIR / name)
(EXECUTED_DIR / "batch_status.json").write_text(json.dumps(batch_status, ensure_ascii=False, indent=2))
print(json.dumps(batch_status, ensure_ascii=False, indent=2))
```

## 最后运行汇总

若部分样本失败，汇总会说明完成情况；评分复现不一致的样本不进入组成部分归因。


```python
summary_path = NOTEBOOK_DIR / "11_Summary_All_10_Datasets.ipynb"
summary_nb = nbformat.read(summary_path, as_version=4)
try:
    NotebookClient(summary_nb, timeout=1800, kernel_name=KERNEL_NAME, resources={"metadata": {"path": str(NOTEBOOK_DIR)}}).execute()
finally:
    nbformat.write(summary_nb, EXECUTED_DIR / summary_path.name)
```
